Kütüphanenin Eklenmesi ve Verinin Okunması

```python
import pandas as pd

df = pd.read_csv('twitter_data.csv')
print("Veriseti boyutu:", df.shape)
```

Verisetinin İncelenmesi

```python
print("Sütun İsimleri:", df.columns.tolist())
print("\nVeri Türleri:", df.dtypes)
df.head()
```

```python
df.info()
df.describe(include='all')
```

Tekrarlanan kayıtların silinmesi

```python
baslangic_kayit_sayisi = df.shape[0]

print("Tekrarlanan kayıt sayısı:", df.duplicated().sum())
df = df.drop_duplicates().copy()
print("Tekrarlanan kayıtlar silindikten sonra boyut:", df.shape)
```

Regex ile Metin Temizleme

```python
import re

def clean_text(text):
    if type(text) == str:
        text = re.sub(r'http\S+', '', text)
        text = re.sub(r'@\w+', '', text)
        text = re.sub(r'#\w+', '', text)
        text = re.sub(r'[^\w\s]', '', text)
        text = text.strip()
    return text

df['clean_text'] = df['text'].apply(clean_text)
df[['text', 'clean_text']].head()
```

Eksik değerlerin giderilmesi

```python
print("Eksik değer sayısı:", df.isnull().sum())
if 'negativereason' in df.columns:
    df['negativereason'] = df['negativereason'].fillna('Not Specified')
df = df.dropna(subset=['clean_text', 'airline_sentiment'])
print("\nEksik veriler işlendikten sonra boyut:", df.shape)
```

Veri Tiplerinin kontrol edilmesi

```python
if 'tweet_created' in df.columns:
    df['tweet_created'] = pd.to_datetime(df['tweet_created'], errors='coerce')
print("Güncel veri tipleri:\n", df.dtypes)
```

```python
son_kayit_sayisi = df.shape[0]
print(f"Temizleme Öncesi Kayıt Sayısı: {baslangic_kayit_sayisi}")
print(f"Temizleme Sonrası Kayıt Sayısı: {son_kayit_sayisi}")
print(f"Toplam {baslangic_kayit_sayisi - son_kayit_sayisi} kayıt pipeline'ın bu adımında temizlendi.")
```

MongoDB Bağlantısı ve Bulk Insert

```python
from pymongo import MongoClient
import pandas as pd

atlas_uri = "ATLAS URI
client = MongoClient(atlas_uri)
db = client['twitter_db']
collection = db['tweetler']
collection.delete_many({})

if 'tweet_created' in df.columns:
    df['tweet_created'] = df['tweet_created'].astype(str)

records = df.to_dict(orient='records')
collection.insert_many(records)

print(f"MongoDB Atlas'a {collection.count_documents({})} adet kayıt başarıyla eklendi!")
```

find() ile Temel Sorgular (3 Adet)

```python
print("1)Sadece 'negative' olan 3 tweeti getir\n")
q1 = collection.find({'airline_sentiment': 'negative'}).limit(3)
for doc in q1: 
    print("-", doc.get('clean_text', ''))

print("\n2)'Virgin America' havayolunun tweetleri\n")
q2 = collection.find({'airline': 'Virgin America'}, {'_id': 0, 'clean_text': 1}).limit(3)
for doc in q2: 
    print("-", doc)

print("\n3)Güven skoru 0.9'dan büyük olan pozitif tweetler\n")
q3 = collection.find({
    'airline_sentiment': 'positive', 
    'airline_sentiment_confidence': {'$gt': 0.9}
}).limit(2)
for doc in q3: 
    print("- Havayolu:", doc.get('airline'), "| Metin:", doc.get('clean_text', ''))
```

Aggregation Pipeline (3 Adet)

```python
print("4. Havayollarına göre toplam tweet sayıları\n")
agg1 = collection.aggregate([{'$group': {'_id': '$airline', 'toplam_tweet': {'$sum': 1}}}])
for doc in agg1: print(doc)

print("\n5. Genel duygu dağılımı (Pozitif/Negatif/Nötr oranları)\n")
agg2 = collection.aggregate([{'$group': {'_id': '$airline_sentiment', 'adet': {'$sum': 1}}}])
for doc in agg2: print(doc)

print("\n6. En çok şikayet edilen 3 negatif sebep\n")
agg3 = collection.aggregate([
    {'$match': {'negativereason': {'$ne': 'Not Specified'}}},
    {'$group': {'_id': '$negativereason', 'frekans': {'$sum': 1}}},
    {'$sort': {'frekans': -1}},
    {'$limit': 3}
])
for doc in agg3: print(doc)
```

İndeks Oluşturma ve Test (1 Adet)

```python
print("7. 'airline' alanına indeks oluşturma ve explain() analizi\n")
collection.create_index('airline')

sorgu_plani = collection.find({'airline': 'United'}).explain()

print("Kullanılan indeks (Winning Plan):")
print(sorgu_plani['queryPlanner']['winningPlan'])
```

UPDATE İşlemleri (2 Adet)

```python
print("8. UPDATE: İlk United Airlines tweetinin retweet sayısını $inc ile 5 artırma\n")

query = {'airline': 'United'}
fields = {'_id': 1,'airline': 1,'retweet_count': 1}
print(list(collection.find(query, fields).limit(1)))

res1 = collection.update_one({'airline': 'United'}, {'$inc': {'retweet_count': 5}})
print(f"Değiştirilen doküman sayısı: {res1.modified_count}")

query = {'airline': 'United'}
fields = {'_id': 1,'airline': 1,'retweet_count': 1}
print(list(collection.find(query, fields).limit(1)))

print("\n9. UPDATE: 'neutral' olan tweetlere $set ile 'is_analyzed' alanı ekleme\n")
res2 = collection.update_many({'airline_sentiment': 'neutral'}, {'$set': {'is_analyzed': True}})
print(f"Güncellenen nötr tweet sayısı: {res2.modified_count}\n")

updated_docs = collection.find({'airline_sentiment': 'neutral', 'is_analyzed': True},{'_id': 0, 'text': 1, 'airline_sentiment': 1, 'is_analyzed': 1}).limit(3)

for doc in updated_docs:
    print(doc)
```

DELETE ve İleri Seviye Serbest Sorgular (3 Adet)

```python
print("10. DELETE: Güven skoru 0.4'ten küçük olan düşük kaliteli verileri silme\n")
silme_islemi = collection.delete_many({'airline_sentiment_confidence': {'$lt': 0.4}})
print(f"Silinen düşük kaliteli kayıt sayısı: {silme_islemi.deleted_count}")

print("\n11. REGEX: İçinde 'delay' veya 'delayed' geçen tweetler\n")
q11 = collection.find({'clean_text': {'$regex': 'delay', '$options': 'i'}}).limit(3)
for doc in q11: 
    print("-", doc.get('clean_text', ''))

print("\n12. İÇ İÇE: Retweet edilmiş ve Eastern Time zone'da atılmış tweetler\n")
q12 = collection.find({'retweet_count': {'$gt': 0}, 'user_timezone': 'Eastern Time (US & Canada)'}).limit(2)
for doc in q12: 
    print(f"- {doc.get('airline')} (RT: {doc.get('retweet_count')})")
```

Dönüşüm (Transformation) İşlemleri (3 Adet)

```python
from pyspark.sql.functions import length

print("1. Transformation: select() ve filter()")
# Sadece United havayoluna ait negatif tweetleri filtreleyip seçme
negatif_united = spark_df.select("airline", "clean_text", "airline_sentiment") \
                         .filter((spark_df.airline == "United") & (spark_df.airline_sentiment == "negative"))
negatif_united.show(3, truncate=False)

print("\n2. Transformation: withColumn()")
# Tweet metninin karakter uzunluğunu yeni bir sütun olarak ekleme
spark_df = spark_df.withColumn("tweet_uzunlugu", length(spark_df.clean_text))
spark_df.select("clean_text", "tweet_uzunlugu").show(3)

print("\n3. Transformation: groupBy()")
# Havayollarına göre toplam tweet sayılarını gruplama
havayolu_grup = spark_df.select("airline").coalesce(1).groupBy("airline").count()
havayolu_grup.show(3)
```
